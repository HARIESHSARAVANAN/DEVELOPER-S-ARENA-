# 1 DATA CLEANING


# 1_data_cleaning

Week 8 Customer Churn & Retention Strategy

## Objective
Inspect, validate, clean, and export the customer churn dataset.

```python
import pandas as pd
df=pd.read_csv("../data/raw_data.csv")
print(df.shape)
display(df.head())
display(df.isna().sum())
print("Duplicate IDs:",df.CustomerID.duplicated().sum())
```

```python
for c in ["Tenure","MonthlyCharges","TotalCharges","SeniorCitizen","Churn"]:
    df[c]=pd.to_numeric(df[c],errors="coerce")
df["CustomerID"]=df["CustomerID"].astype(str).str.strip()
df=df.drop_duplicates("CustomerID").dropna().reset_index(drop=True)
df.to_csv("../data/cleaned_data.csv",index=False)
print(df.shape)
```

# 3 ANALYSIS


# 3_analysis

Week 8 Customer Churn & Retention Strategy

## Objective
Apply statistical tests and build a baseline churn classification model.

```python
from scipy.stats import ttest_ind, chi2_contingency
print("Tenure p:",ttest_ind(df.loc[df.Churn==0,"Tenure"],df.loc[df.Churn==1,"Tenure"],equal_var=False).pvalue)
print("Monthly Charges p:",ttest_ind(df.loc[df.Churn==0,"MonthlyCharges"],df.loc[df.Churn==1,"MonthlyCharges"],equal_var=False).pvalue)
for c in ["Contract","PaymentMethod","PaperlessBilling"]:
    print(c,chi2_contingency(pd.crosstab(df[c],df.Churn))[1])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report,roc_auc_score
X=df.drop(columns=["CustomerID","Churn","ChurnLabel"]); y=df.Churn
cat=["Contract","PaymentMethod","PaperlessBilling"]; num=["Tenure","MonthlyCharges","TotalCharges","SeniorCitizen"]
pre=ColumnTransformer([("num",StandardScaler(),num),("cat",OneHotEncoder(handle_unknown="ignore"),cat)])
pipe=Pipeline([("pre",pre),("model",LogisticRegression(max_iter=2000,class_weight="balanced",random_state=42))])
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,stratify=y,random_state=42)
pipe.fit(Xtr,ytr); p=pipe.predict(Xte); prob=pipe.predict_proba(Xte)[:,1]
print(classification_report(yte,p)); print("ROC-AUC:",roc_auc_score(yte,prob))
```

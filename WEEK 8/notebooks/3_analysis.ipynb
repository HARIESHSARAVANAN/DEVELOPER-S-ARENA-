```
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report,roc_auc_score
X=df.drop(columns=["CustomerID","Churn","ChurnLabel"]); y=df.Churn
cat=["Contract","PaymentMethod","PaperlessBilling"]; num=["Tenure","MonthlyCharges","TotalCharges","SeniorCitizen"]
pre=ColumnTransformer([("num",StandardScaler(),num),("cat",OneHotEncoder(handle_unknown="ignore"),cat)])
pipe=Pipeline([("pre",pre),("model",LogisticRegression(max_iter=2000,class_weight="balanced",random_state=42))])
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,stratify=y,random_state=42)
pipe.fit(Xtr,ytr); p=pipe.predict(Xte); prob=pipe.predict_proba(Xte)[:,1]
print(classification_report(yte,p)); print("ROC-AUC:",roc_auc_score(yte,prob))
```

Week 8 Customer Churn & Retention Strategy

## Objective
Inspect, validate, clean, and export the customer churn dataset.

```python
import pandas as pd
df=pd.read_csv("../data/raw_data.csv")
print(df.shape)
display(df.head())
display(df.isna().sum())
print("Duplicate IDs:",df.CustomerID.duplicated().sum())
```

```python
for c in ["Tenure","MonthlyCharges","TotalCharges","SeniorCitizen","Churn"]:
    df[c]=pd.to_numeric(df[c],errors="coerce")
df["CustomerID"]=df["CustomerID"].astype(str).str.strip()
df=df.drop_duplicates("CustomerID").dropna().reset_index(drop=True)
df.to_csv("../data/cleaned_data.csv",index=False)
print(df.shape)
```

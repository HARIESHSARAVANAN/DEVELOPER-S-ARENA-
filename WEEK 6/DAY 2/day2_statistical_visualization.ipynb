```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("sales_data.csv")

df["Date"] = pd.to_datetime(df["Date"])

print(df.head())
```

```python
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Product",
    y="Price"
)

plt.title("Price Distribution by Product")
plt.xlabel("Product")
plt.ylabel("Price")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    "visualizations/price_boxplot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x="Product",
    y="Total_Sales"
)

plt.title("Total Sales Distribution by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    "visualizations/sales_violinplot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Region",
    y="Total_Sales"
)

plt.title("Total Sales Distribution by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.tight_layout()

plt.savefig(
    "visualizations/regional_sales_boxplot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
numeric_data = df.select_dtypes(include="number")

correlation_matrix = numeric_data.corr()

print("\nCORRELATION MATRIX")
print(correlation_matrix)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.savefig(
    "visualizations/correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

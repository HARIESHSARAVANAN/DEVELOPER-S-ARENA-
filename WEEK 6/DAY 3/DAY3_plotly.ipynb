```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv("sales_data.csv")

df["Date"] = pd.to_datetime(df["Date"])

print(df.head())
```

```python
daily_sales = (
    df.groupby("Date")["Total_Sales"]
    .sum()
    .reset_index()
)

fig = px.line(
    daily_sales,
    x="Date",
    y="Total_Sales",
    title="Interactive Daily Sales Trend",
    markers=True
)

fig.update_layout(
    xaxis_title="Date",
    yaxis_title="Total Sales"
)

fig.show()
```

```python
region_sales = (
    df.groupby("Region")["Total_Sales"]
    .sum()
    .reset_index()
    .sort_values("Total_Sales", ascending=False)
)

fig = px.bar(
    region_sales,
    x="Region",
    y="Total_Sales",
    title="Interactive Regional Sales",
    hover_data=["Total_Sales"]
)

fig.update_layout(
    xaxis_title="Region",
    yaxis_title="Total Sales"
)

fig.show()
```

```python
fig = px.scatter(
    df,
    x="Quantity",
    y="Total_Sales",
    color="Product",
    hover_data=[
        "Date",
        "Product",
        "Quantity",
        "Price",
        "Region"
    ],
    title="Quantity vs Total Sales"
)

fig.update_layout(
    xaxis_title="Quantity",
    yaxis_title="Total Sales"
)

fig.show()
```

```python
product_region_sales = (
    df.groupby(["Product", "Region"])["Total_Sales"]
    .sum()
    .reset_index()
)

fig = px.bar(
    product_region_sales,
    x="Product",
    y="Total_Sales",
    color="Region",
    barmode="group",
    hover_data=["Region", "Total_Sales"],
    title="Product Sales by Region"
)

fig.update_layout(
    xaxis_title="Product",
    yaxis_title="Total Sales"
)

fig.show()
```

```python
fig = px.bar(
    df,
    x="Product",
    y="Total_Sales",
    color="Region",
    title="Interactive Sales by Product"
)

fig.update_layout(
    xaxis_title="Product",
    yaxis_title="Total Sales"
)

fig.show()
```

```python
fig.write_html(
    "visualizations/interactive_sales_trend.html"
)
```

```python
fig.write_html(
    "visualizations/interactive_product_sales.html"
)
```

```python
fig.write_html(
    "visualizations/interactive_regional_sales.html"
)
```
